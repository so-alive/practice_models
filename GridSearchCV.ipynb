```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split

X = df.drop('species', axis = 1)
y = df['species']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors=5)
model_knn.fit(X_train, y_train)
model_knn.score(X_test, y_test)
```

```python
from sklearn.svm import SVC

model_svm = SVC(gamma = 'auto', C=30, kernel='linear')
model_svm.fit(X_train, y_train)
model_svm.score(X_test, y_test)
```

# Grid Search CV

```python
from sklearn.model_selection import GridSearchCV

classifier = GridSearchCV((model_svm), {
    'C' : [1,10,20,30],
    'kernel' : ['rbf', 'linear']
}, cv = 5, return_train_score = False)

classifier.fit(X, y)
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
results
```

```python
results[['param_C', 'param_kernel', 'mean_test_score']]
```

```python
classifier = GridSearchCV((model_knn), {
    'n_neighbors' : [1,5,10,20],
    'weights' : ['uniform','distance'],
    'p' : [1,2]
}, cv = 5, return_train_score = False)

classifier.fit(X,y)
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
results
```

```python
results[['param_n_neighbors', 'param_p', 'param_weights','mean_test_score']]
```

```python
from sklearn.model_selection import RandomizedSearchCV

classifier_random = RandomizedSearchCV((model_svm), {
    'C' : [1,10,20,30],
    'kernel' : ['rbf', 'linear']
},n_iter = 4, cv = 5, return_train_score=False)

classifier_random.fit(X, y)
classifier_random.cv_results_
```

```python
results_random = pd.DataFrame(classifier_random.cv_results_)
results_random
```

```python
results_random[['param_kernel', 'param_C', 'mean_test_score']]
```
